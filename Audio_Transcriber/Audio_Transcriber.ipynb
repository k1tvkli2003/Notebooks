## 📊 Step 3: گزارش نهایی

```python
# بارگذاری تابع format_time از ماژول پردازش
from audio_transcriber_processor import format_time

print("="*80)
print("📊 گزارش کامل پردازش")
print("="*80)
print()

# بررسی وجود results
if 'results' not in globals():
    print("❌ هنوز پردازشی انجام نشده است!")
    print("💡 لطفاً ابتدا Step 2 را اجرا کنید")
else:
    # فایل‌های موفق
    if results['success']:
        print(f"✅ فایل‌های موفق ({len(results['success'])}):\n")
        for item in results['success'][:20]:
            from pathlib import Path
            filename = Path(item['file']).name
            cached_text = " (از کش)" if item.get('cached') else ""
            print(f"   ✓ {filename}{cached_text}")
            if not item.get('cached'):
                duration_sec = item.get('duration', 0)
                duration_min = int(duration_sec // 60)
                duration_s = int(duration_sec % 60)
                print(f"      ⏱️ {format_time(item['time'])} | 🎵 {duration_min}:{duration_s:02d} دقیقه")
        if len(results['success']) > 20:
            print(f"\n   ... و {len(results['success']) - 20} فایل دیگر")
        print()
    
    # فایل‌های ناموفق
    if results['failed']:
        print(f"❌ فایل‌های ناموفق ({len(results['failed'])}):\n")
        for item in results['failed']:
            from pathlib import Path
            filename = Path(item['file']).name
            print(f"   ✗ {filename}")
            print(f"      خطا: {item['error'][:150]}...")
            print()
    
    print("="*80)
    print(f"📁 همه خروجی‌ها در کنار فایل‌های صوتی ذخیره شده‌اند")
    print("="*80)
```

## 🗑️ Step 4: حذف فایل‌های تولید شده

**⚠️ توجه:** این سل همه فایل‌های `.md` تولید شده در مرحله پردازش را حذف می‌کند!

```python
import os
from pathlib import Path

def cleanup_generated_files(files):
    """حذف فایل‌های md تولید شده"""
    deleted_count = 0
    errors = []
    
    print("🗑️ در حال حذف فایل‌های تولید شده...")
    print()
    
    for audio_file in files:
        md_path = audio_file.parent / f"{audio_file.stem}.md"
        
        if md_path.exists():
            try:
                os.remove(md_path)
                print(f"   ✓ حذف شد: {md_path.name}")
                deleted_count += 1
            except Exception as e:
                errors.append(f"{md_path.name}: {e}")
                print(f"   ✗ خطا: {md_path.name}")
    
    print()
    print("="*80)
    print(f"📊 خلاصه: {deleted_count} فایل حذف شد")
    if errors:
        print(f"⚠️ {len(errors)} خطا رخ داد")
    print("="*80)
    
    return {'deleted': deleted_count, 'errors': errors}

# اجرای حذف
if not audio_files:
    print("❌ هیچ فایلی برای حذف یافت نشد!")
    print("💡 لطفاً ابتدا Step 1 را اجرا کنید")
else:
    cleanup_result = cleanup_generated_files(audio_files)
```
