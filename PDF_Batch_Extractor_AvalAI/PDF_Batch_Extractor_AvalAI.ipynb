```python
# نصب پکیج‌های مورد نیاز
import subprocess, sys

_required_packages = {
    "nest_asyncio": "nest_asyncio",
    "ipywidgets": "ipywidgets",
    "PyPDF2": "PyPDF2",
    "httpx": "httpx",
    "fitz": "PyMuPDF",
    "docx": "python-docx",
}

print("📦 بررسی و نصب پکیج‌های مورد نیاز...")
for _import_name, _pip_name in _required_packages.items():
    try:
        __import__(_import_name)
    except ImportError:
        print(f"   ⬇️ نصب {_pip_name}...")
        subprocess.check_call([sys.executable, "-m", "pip", "install", _pip_name, "-q"])
print("✅ همه پکیج‌ها آماده‌اند!\n")

import nest_asyncio
nest_asyncio.apply()

# بارگذاری تمام تنظیمات و UI از ماژول خارجی
from avalai_ocr_setup import *

# نمایش وضعیت فایل‌های انتخاب شده
if selected_pdf_files:
    print()
    print("="*80)
    print(f"✅ {len(selected_pdf_files)} فایل آماده پردازش")
    total_size = sum(pdf.stat().st_size for pdf in selected_pdf_files) / (1024 * 1024)
    print(f"💾 حجم کل: {total_size:.2f} MB")
    print(f"🚀 به Step 2 بروید برای شروع پردازش!")
    print("="*80)
```

## 🚀 Step 2: پردازش فایل‌ها با AvalAI OCR

**⚡ مدل و فرمت خروجی انتخاب شده از Step 1 استفاده می‌شوند**

**📝 توجه:** خروجی مستقیماً از AvalAI API استخراج و در فرمت‌های انتخابی ذخیره می‌شود

```python
# بارگذاری ماژول پردازش
from avalai_ocr_processor import process_pdfs
from avalai_ocr_setup import selected_pdf_files, selected_model, extract_images

# اجرای پردازش
if not selected_pdf_files:
    print("❌ هیچ فایلی انتخاب نشده است!")
    print("💡 لطفاً به Step 1 برگردید و فایل‌ها را انتخاب کنید")
else:
    print(f"🤖 مدل انتخاب شده: {selected_model}")
    print(f"🖼️  استخراج تصاویر: {'✅ فعال' if extract_images else '❌ غیرفعال'}")
    print(f"📄 تعداد فایل‌ها: {len(selected_pdf_files)}")
    print()
    results = await process_pdfs(selected_pdf_files, model_id=selected_model)
```

## 📊 Step 3: گزارش نهایی

```python
from avalai_ocr_processor import format_time

print("="*80)
print("📊 گزارش کامل پردازش")
print("="*80)
print()

if 'results' not in globals():
    print("❌ هنوز پردازشی انجام نشده است!")
    print("💡 لطفاً ابتدا Step 2 را اجرا کنید")
else:
    if results['success']:
        print(f"✅ فایل‌های موفق ({len(results['success'])}):\n")
        for item in results['success'][:20]:
            from pathlib import Path
            filename = Path(item['file']).name
            img_info = f" | 🖼️ {item.get('images', 0)} تصویر" if item.get('images', 0) > 0 else ""
            print(f"   ✓ {filename}")
            print(f"      ⏱️  {format_time(item['time'])} | 📄 {item.get('pages', 0)} صفحه{img_info}")
        if len(results['success']) > 20:
            print(f"\n   ... و {len(results['success']) - 20} فایل دیگر")
        print()

    if results['failed']:
        print(f"❌ فایل‌های ناموفق ({len(results['failed'])}):\n")
        for item in results['failed']:
            from pathlib import Path
            filename = Path(item['file']).name
            print(f"   ✗ {filename}")
            print(f"      خطا: {item['error'][:150]}...")
            print()

    total_images = sum(r.get('images', 0) for r in results['success'])
    if total_images > 0:
        print(f"🖼️  مجموع تصاویر استخراج شده: {total_images}")

    print("="*80)
    print(f"📁 فایل‌های .md در کنار PDF‌ها ذخیره شده‌اند")
    print("="*80)
```
